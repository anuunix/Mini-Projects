```python
def getdate():
    import datetime
    return datetime.datetime.now()

def retrieve_Anuska():
    Type=int(input("For Excercise 1, For diet:2"))
    if Type==1:
        f=open("Anuska.excer","r")
        content=f.read()
        print(f"Your Content for date {getdate()} is {content}")
    elif Type==2:
        f=open("Anuska.diet","r")
        content=f.read()
        print(f"Your Content for date {getdate()} is {content}")

def retrieve_Deepak():
    Type=int(input("For Excercise 1, For diet:2"))
    if Type==1:
        f=open("Deepak.excer","r")
        content=f.read()
        print(f"Your Content for date {getdate()} is {content}")
    elif Type==2:
        f=open("Deepak.diet","r")
        content=f.read()
        print(f"Your Content for date {getdate()} is {content}")  

def Log_Anuska():
    Type=int(input("For Excercise 1, For diet:2"))
    if Type==1:
        f=open("Anuska.excer","a")
        content=input("Enter your input /n")
        f.write(str([str(getdate())])+": "+content+"\n")
    elif Type==2:
        f=open("Anuska.diet","a")
        content=input("Enter your input /n")
        f.write(str([str(getdate())])+": "+content+"\n")

def Log_Deepak():
    Type=int(input("For Excercise 1, For diet:2"))
    if Type==1:
        f=open("Deepak.excer","a")
        content=input("Enter your input /n")
        f.write(str([str(getdate())])+": "+content+"\n")
    elif Type==2:
        f=open("Deepak.diet","a")
        content=input("Enter your input /n")
        f.write(str([str(getdate())])+": "+content+"\n")


while True:
    Name=input("Enter your name:")
    Operation=input("Enter R to retrive and L for log")
    
    if Name=='Anuska' and Operation=='R':
        retrieve_Anuska()
    elif Name=='Anuska' and Operation=='L':
        Log_Anuska()
    elif Name == 'Deepak' and Operation == 'R':
        retrieve_Deepak()
    elif Name == 'Deepak' and Operation == 'L':
        Log_Deepak()
    else:
        print("No matching case") 
    

    Next = input("Do you want to continue? Y/n: ").strip().upper()
    if Next != 'Y':
        break
```
