```python
class library:
    def __init__(self):
        self.books = []

    def add_book(self):
        new_books = input("Enter list of books separated by commas: ")
        book_list = [book.strip() for book in new_books.split(",")]
        self.books.extend(book_list)
        with open("Book.txt", "a") as f:
            f.write(", ".join(book_list) + "\n")  # To convert list to a string and append to file
        return f"Books added: {book_list}"
    
    def display_books(self):
        with open("Book.txt", "r") as f:
            print(f.read())

    def lend_book(self):
        Diction = {}
        number = int(input("Enter number of users you want to add: "))
        for item in range(number):
            name = input("Enter your name: ")
            book_name = input("Enter book name: ")
            Diction[name] = book_name
            print(f"\n{book_name} taken by {name}")
        with open("LendBook.txt", "a") as f:
            for name, book in Diction.items():
                f.write(f"{name} borrowed {book}\n")

    def search_books(self):
        book_name = input("Enter book name you want to search: ").strip()
        with open("Book.txt", "r") as f:
            books = f.read()
        with open("LendBook.txt", "r") as f:
            lbooks = f.readlines()  # Read lines instead of the entire content
        
        # Check if the book is available in Book.txt
        if book_name.lower() in books.lower():
            print("Book is available")
            lend = input("Enter 'l' if you want to lend the book: ").strip().lower()
            if lend == 'l':
                self.lend_book()
        # Check if the book is already borrowed in LendBook.txt
        else:
            found = False
            for line in lbooks:
                if book_name.lower() in line.lower():
                    print(line.strip())  # Print the line (single entry) where the book is borrowed
                    found = True
                    break  # Stop after finding the first match

            if not found:
                print("Book not found in the borrowed books list.")

# Main code with loop for continuous input
def main():
    anuska_library = library()

    while True:
        work = input("Enter what do you want to do:\n1: To display type 'd'\n2: To lend type 'l'\n3: To add type 'a'\n4: To search for books type 's'\nOr type 'q' to quit: ").strip().lower()

        if work == 'q':
            print("Goodbye!")
            break  # Exit the loop if 'q' is typed
        elif work == 'a':
            print(anuska_library.add_book())
        elif work == 'd':
            anuska_library.display_books()
        elif work == 'l':
            anuska_library.lend_book()
        elif work == 's':
            anuska_library.search_books()
        else:
            print("Invalid choice. Try again.")

# Start the program
main()
```
