```python
import random

type = ["s", "w", "g"]

while True:
    comp = random.choice(type)
    user = input("Enter your choice (s: Snake, w: Water, g: Gun): ").lower()

    print(f"Computer chose: {comp}")

    if user == comp:
        print("It's a tie!")
    elif user == 'g' and comp == 'w':
        print("You win!")
    elif user == 'w' and comp == 'g':
        print("You lose!")
    elif user == 's' and comp == 'g':
        print("You lose!")
    elif user == 'g' and comp == 's':
        print("You win!")
    elif user == 'w' and comp == 's':
        print("You win!")
    elif user == 's' and comp == 'w':
        print("You lose!")
    else:
        print("Invalid input!")

    again = input("Do you want to play again? (y/n): ").lower()
    if again != 'y':
        break
```
